# Regression Algorithm

This notebook contains:
- Linear Regression
- Ridge Regression
- Lasso Regression
- Gradient Boosting Regression
- Gradient Boosting Hyperparameter Tuning
- Evaluation using R², RMSE, and MAE
- Actual vs Predicted plots
- Residual plots
- Gradient Boosting feature importance

```python
# If required, install dependencies in the VS Code terminal:
# pip install pandas numpy matplotlib scikit-learn openpyxl

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

print("Libraries imported successfully.")
```

## 1. Load / Prepare Data

This section expects the following variables to already exist:
- `X_train`
- `X_test`
- `y_train`
- `y_test`
- `X_train_scaled`
- `X_test_scaled`
- `final_features`

If you have not created these variables yet, replace the example loading/preprocessing section with your dataset-specific preprocessing code.

```python
# ============================================================
# DATA CHECK
# ============================================================
# Run this cell after your preprocessing code has created:
# X_train, X_test, y_train, y_test
# X_train_scaled, X_test_scaled
# final_features

required_variables = [
    "X_train", "X_test", "y_train", "y_test",
    "X_train_scaled", "X_test_scaled", "final_features"
]

missing = [var for var in required_variables if var not in globals()]

if missing:
    print("Missing variables:", missing)
    print("\nCreate these variables in your preprocessing cells before running the models.")
else:
    print("All required variables are available.")
    print("X_train shape:", X_train.shape)
    print("X_test shape :", X_test.shape)
    print("y_train shape:", np.shape(y_train))
    print("y_test shape :", np.shape(y_test))
```

## 2. Linear Regression

```python
linear_model = LinearRegression()

linear_model.fit(
    X_train_scaled,
    y_train
)

y_train_pred_linear = linear_model.predict(
    X_train_scaled
)

y_test_pred_linear = linear_model.predict(
    X_test_scaled
)
```

```python
train_r2_linear = r2_score(
    y_train,
    y_train_pred_linear
)

test_r2_linear = r2_score(
    y_test,
    y_test_pred_linear
)

train_rmse_linear = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred_linear
    )
)

test_rmse_linear = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred_linear
    )
)

train_mae_linear = mean_absolute_error(
    y_train,
    y_train_pred_linear
)

test_mae_linear = mean_absolute_error(
    y_test,
    y_test_pred_linear
)

print("Linear Regression")
print("----------------------------")
print("Train R²  :", train_r2_linear)
print("Test R²   :", test_r2_linear)
print("Train RMSE:", train_rmse_linear)
print("Test RMSE :", test_rmse_linear)
print("Train MAE :", train_mae_linear)
print("Test MAE  :", test_mae_linear)

linear_coefficients = pd.DataFrame({
    "Feature": final_features,
    "Coefficient": linear_model.coef_
})

display(linear_coefficients)

print("Intercept:", linear_model.intercept_)
```

```python
# Actual vs Predicted Values

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_test_pred_linear,
    alpha=0.6
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.xlabel("Actual Sound Pressure Level (dB)")
plt.ylabel("Predicted Sound Pressure Level (dB)")
plt.title("Linear Regression: Actual vs Predicted")

plt.tight_layout()
plt.show()

# Residual Plot
residuals_linear = y_test - y_test_pred_linear

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_pred_linear,
    residuals_linear,
    alpha=0.6
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Sound Pressure Level (dB)")
plt.ylabel("Residuals (Actual - Predicted)")
plt.title("Linear Regression: Residual Plot")

plt.tight_layout()
plt.show()
```

## 3. Ridge Regression

```python
ridge_model = Ridge(alpha=1.0)

ridge_model.fit(
    X_train_scaled,
    y_train
)

y_train_pred_ridge = ridge_model.predict(
    X_train_scaled
)

y_test_pred_ridge = ridge_model.predict(
    X_test_scaled
)

train_r2_ridge = r2_score(
    y_train,
    y_train_pred_ridge
)

test_r2_ridge = r2_score(
    y_test,
    y_test_pred_ridge
)

train_rmse_ridge = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred_ridge
    )
)

test_rmse_ridge = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred_ridge
    )
)

train_mae_ridge = mean_absolute_error(
    y_train,
    y_train_pred_ridge
)

test_mae_ridge = mean_absolute_error(
    y_test,
    y_test_pred_ridge
)

print("Ridge Regression")
print("----------------------------")
print("Train R²  :", train_r2_ridge)
print("Test R²   :", test_r2_ridge)
print("Train RMSE:", train_rmse_ridge)
print("Test RMSE :", test_rmse_ridge)
print("Train MAE :", train_mae_ridge)
print("Test MAE  :", test_mae_ridge)

ridge_coefficients = pd.DataFrame({
    "Feature": final_features,
    "Coefficient": ridge_model.coef_
})

display(ridge_coefficients)
```

## 4. Lasso Regression

```python
lasso_model = Lasso(alpha=0.1)

lasso_model.fit(
    X_train_scaled,
    y_train
)

y_train_pred_lasso = lasso_model.predict(
    X_train_scaled
)

y_test_pred_lasso = lasso_model.predict(
    X_test_scaled
)

train_r2_lasso = r2_score(
    y_train,
    y_train_pred_lasso
)

test_r2_lasso = r2_score(
    y_test,
    y_test_pred_lasso
)

train_rmse_lasso = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred_lasso
    )
)

test_rmse_lasso = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred_lasso
    )
)

train_mae_lasso = mean_absolute_error(
    y_train,
    y_train_pred_lasso
)

test_mae_lasso = mean_absolute_error(
    y_test,
    y_test_pred_lasso
)

print("Lasso Regression")
print("----------------------------")
print("Train R²  :", train_r2_lasso)
print("Test R²   :", test_r2_lasso)
print("Train RMSE:", train_rmse_lasso)
print("Test RMSE :", test_rmse_lasso)
print("Train MAE :", train_mae_lasso)
print("Test MAE  :", test_mae_lasso)

lasso_coefficients = pd.DataFrame({
    "Feature": final_features,
    "Coefficient": lasso_model.coef_
})

display(lasso_coefficients)
```

## 5. Gradient Boosting Regression

```python
gb_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

gb_model.fit(
    X_train,
    y_train
)

y_train_pred_gb = gb_model.predict(
    X_train
)

y_test_pred_gb = gb_model.predict(
    X_test
)

train_r2_gb = r2_score(
    y_train,
    y_train_pred_gb
)

test_r2_gb = r2_score(
    y_test,
    y_test_pred_gb
)

train_rmse_gb = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred_gb
    )
)

test_rmse_gb = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred_gb
    )
)

train_mae_gb = mean_absolute_error(
    y_train,
    y_train_pred_gb
)

test_mae_gb = mean_absolute_error(
    y_test,
    y_test_pred_gb
)

print("Gradient Boosting Regression")
print("----------------------------")
print("Train R²  :", train_r2_gb)
print("Test R²   :", test_r2_gb)
print("Train RMSE:", train_rmse_gb)
print("Test RMSE :", test_rmse_gb)
print("Train MAE :", train_mae_gb)
print("Test MAE  :", test_mae_gb)
```

```python
# Actual vs Predicted

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_test_pred_gb,
    alpha=0.6
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.xlabel("Actual Sound Pressure Level (dB)")
plt.ylabel("Predicted Sound Pressure Level (dB)")
plt.title("Gradient Boosting: Actual vs Predicted")

plt.tight_layout()
plt.show()

# Residual Plot

residuals_gb = y_test - y_test_pred_gb

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_pred_gb,
    residuals_gb,
    alpha=0.6
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Sound Pressure Level (dB)")
plt.ylabel("Residuals (Actual - Predicted)")
plt.title("Gradient Boosting: Residual Plot")

plt.tight_layout()
plt.show()
```

## 6. Gradient Boosting Hyperparameter Tuning

```python
gb = GradientBoostingRegressor(
    random_state=42
)

param_grid_gb = {
    "n_estimators": [100, 200, 300],
    "learning_rate": [0.03, 0.05, 0.1],
    "max_depth": [2, 3, 4],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

gb_grid = GridSearchCV(
    estimator=gb,
    param_grid=param_grid_gb,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

gb_grid.fit(
    X_train,
    y_train
)

print("Best Parameters:")
print(gb_grid.best_params_)

print("\nBest Cross-Validation RMSE:")
print(-gb_grid.best_score_)

best_gb = gb_grid.best_estimator_

y_train_pred_gb_tuned = best_gb.predict(
    X_train
)

y_test_pred_gb_tuned = best_gb.predict(
    X_test
)
```

```python
print("Tuned Gradient Boosting Regression")
print("----------------------------")

print(
    "Best Parameters:",
    gb_grid.best_params_
)

print(
    "Train R²:",
    r2_score(
        y_train,
        y_train_pred_gb_tuned
    )
)

print(
    "Test R²:",
    r2_score(
        y_test,
        y_test_pred_gb_tuned
    )
)

print(
    "Train RMSE:",
    np.sqrt(
        mean_squared_error(
            y_train,
            y_train_pred_gb_tuned
        )
    )
)

print(
    "Test RMSE:",
    np.sqrt(
        mean_squared_error(
            y_test,
            y_test_pred_gb_tuned
        )
    )
)

print(
    "Train MAE:",
    mean_absolute_error(
        y_train,
        y_train_pred_gb_tuned
    )
)

print(
    "Test MAE:",
    mean_absolute_error(
        y_test,
        y_test_pred_gb_tuned
    )
)
```

```python
# Feature Importance

gb_importance = pd.DataFrame({
    "Feature": final_features,
    "Importance": best_gb.feature_importances_
})

gb_importance = gb_importance.sort_values(
    by="Importance",
    ascending=False
)

display(gb_importance)

plt.figure(figsize=(9, 6))

plt.barh(
    gb_importance["Feature"],
    gb_importance["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("Gradient Boosting Feature Importance")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Tuned Gradient Boosting: Actual vs Predicted

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_test_pred_gb_tuned,
    alpha=0.6
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.xlabel("Actual Sound Pressure Level (dB)")
plt.ylabel("Predicted Sound Pressure Level (dB)")
plt.title("Tuned Gradient Boosting: Actual vs Predicted")

plt.tight_layout()
plt.show()

# Tuned Gradient Boosting: Residual Plot

residuals_gb_tuned = (
    y_test - y_test_pred_gb_tuned
)

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_pred_gb_tuned,
    residuals_gb_tuned,
    alpha=0.6
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Sound Pressure Level (dB)")
plt.ylabel("Residuals (Actual - Predicted)")
plt.title("Tuned Gradient Boosting: Residual Plot")

plt.tight_layout()
plt.show()
```

## 7. Model Comparison

This table collects the test-set metrics from all four regression models.

**Interpretation:**
- Higher **R²** is generally better.
- Lower **RMSE** is better.
- Lower **MAE** is better.

```python
comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression",
        "Gradient Boosting",
        "Tuned Gradient Boosting"
    ],
    "Test R²": [
        test_r2_linear,
        test_r2_ridge,
        test_r2_lasso,
        test_r2_gb,
        r2_score(y_test, y_test_pred_gb_tuned)
    ],
    "Test RMSE": [
        test_rmse_linear,
        test_rmse_ridge,
        test_rmse_lasso,
        test_rmse_gb,
        np.sqrt(mean_squared_error(y_test, y_test_pred_gb_tuned))
    ],
    "Test MAE": [
        test_mae_linear,
        test_mae_ridge,
        test_mae_lasso,
        test_mae_gb,
        mean_absolute_error(y_test, y_test_pred_gb_tuned)
    ]
})

display(comparison.sort_values(by="Test R²", ascending=False).reset_index(drop=True))
```
