# Classification Algorithm

## Logistic Regression

This notebook contains:
- Logistic Regression
- Accuracy, Precision, Recall, F1-score
- Classification report
- Confusion matrix
- Training vs testing accuracy
- ROC-AUC (One-vs-Rest, weighted)

Open this `.ipynb` file in VS Code and select a Python/Jupyter kernel.

```python
# Required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

print("Libraries imported successfully.")
```

## Data Requirements

Run your preprocessing/data-loading cells before this notebook section.

Required variables:
- `X_train`
- `X_test`
- `y_train`
- `y_test`
- `X_train_scaled`
- `X_test_scaled`

Target classes:
- Normal
- Suspect
- Pathologic

```python
# Check required variables
required_variables = [
    "X_train", "X_test", "y_train", "y_test",
    "X_train_scaled", "X_test_scaled"
]

missing = [v for v in required_variables if v not in globals()]

if missing:
    print("Missing variables:", missing)
    print("Run your data loading and preprocessing cells first.")
else:
    print("All required variables are available.")
    print("X_train shape:", X_train.shape)
    print("X_test shape :", X_test.shape)
    print("y_train shape:", np.shape(y_train))
    print("y_test shape :", np.shape(y_test))
```

## 1. Logistic Regression

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(
    X_train_scaled,
    y_train
)

y_train_pred_logistic = logistic_model.predict(X_train_scaled)
y_test_pred_logistic = logistic_model.predict(X_test_scaled)

print("Logistic Regression model trained successfully.")
```

## 2. Accuracy, Precision, Recall and F1-score

```python
accuracy_logistic = accuracy_score(
    y_test,
    y_test_pred_logistic
)

precision_logistic = precision_score(
    y_test,
    y_test_pred_logistic,
    average="macro"
)

recall_logistic = recall_score(
    y_test,
    y_test_pred_logistic,
    average="macro"
)

f1_logistic = f1_score(
    y_test,
    y_test_pred_logistic,
    average="macro"
)

print("Logistic Regression Results")
print("---------------------------")
print(f"Accuracy : {accuracy_logistic:.4f}")
print(f"Precision: {precision_logistic:.4f}")
print(f"Recall   : {recall_logistic:.4f}")
print(f"F1-score : {f1_logistic:.4f}")
```

## 3. Classification Report

```python
print(
    classification_report(
        y_test,
        y_test_pred_logistic,
        target_names=[
            "Normal",
            "Suspect",
            "Pathologic"
        ]
    )
)
```

## 4. Confusion Matrix

```python
cm_logistic = confusion_matrix(
    y_test,
    y_test_pred_logistic
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_logistic,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Normal", "Suspect", "Pathologic"],
    yticklabels=["Normal", "Suspect", "Pathologic"]
)

plt.title("Logistic Regression — Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()
```

## 5. Training vs Testing Accuracy

```python
train_accuracy_logistic = accuracy_score(
    y_train,
    y_train_pred_logistic
)

test_accuracy_logistic = accuracy_score(
    y_test,
    y_test_pred_logistic
)

print(f"Training Accuracy: {train_accuracy_logistic:.4f}")
print(f"Testing Accuracy : {test_accuracy_logistic:.4f}")
```

## 6. ROC-AUC — One-vs-Rest

```python
roc_auc_logistic = roc_auc_score(
    y_test,
    logistic_model.predict_proba(X_test_scaled),
    multi_class="ovr",
    average="weighted"
)

print(
    "ROC-AUC (OvR, weighted) — Logistic Regression: "
    f"{roc_auc_logistic:.4f}"
)
```

## 7. Final Metrics Summary

```python
logistic_summary = pd.DataFrame({
    "Metric": [
        "Training Accuracy",
        "Testing Accuracy",
        "Precision (Macro)",
        "Recall (Macro)",
        "F1-score (Macro)",
        "ROC-AUC (OvR, Weighted)"
    ],
    "Score": [
        train_accuracy_logistic,
        test_accuracy_logistic,
        precision_logistic,
        recall_logistic,
        f1_logistic,
        roc_auc_logistic
    ]
})

display(logistic_summary)
```
